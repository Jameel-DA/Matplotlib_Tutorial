```python
# froma pandas dataframe
bastman = pd.read_csv('sharma-kohli.csv')
bastman
```

```python
plt.plot(bastman['index'],bastman['V Kohli'])
plt.show()
```

```python
# plotting multiple plots
plt.plot(bastman['index'],bastman['V Kohli'])
plt.plot(bastman['index'],bastman['RG Sharma'])
```

```python
# label title
plt.plot(bastman['index'],bastman['V Kohli'])
plt.plot(bastman['index'],bastman['RG Sharma'])
plt.title('Rohit Sharma v/s Virat Kohli carrer comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# colors(hex) and line(width and style) and marker(size)
plt.plot(bastman['index'],bastman['V Kohli'], color = "black", linestyle='dotted', linewidth=1, marker="o", markersize=10)
plt.plot(bastman['index'],bastman['RG Sharma'],color = "blue", linestyle='dashed',linewidth=2, marker="D",markersize=5)
plt.title('Rohit Sharma v/s Virat Kohli carrer comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# legend -> location
plt.plot(bastman['index'],bastman['V Kohli'], color = "black", linestyle='dotted', linewidth=1, marker="o", markersize=10, label='V Kohli')
plt.plot(bastman['index'],bastman['RG Sharma'],color = "blue", linestyle='dashed',linewidth=2, marker="D",markersize=5, label='RG Sharma')
plt.title('Rohit Sharma v/s Virat Kohli carrer comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.legend(loc='right') #loc set the location of legend
```

```python
# limiting axes
price = [4000,54000,57000,49000,47000,4500000]
year = [2015,2016,2017,2018,2019,2020]
plt.plot(year,price)
plt.ylim(0,100000) 
plt.xlim(2016,2018) 
```

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 35]

# Plot
plt.plot(x, y, marker='o')

# Customizing xticks and yticks
plt.xticks(
    ticks=[1, 2, 3, 4, 5],  # Locations of the ticks
    labels=['One', 'Two', 'Three', 'Four', 'Five'],  # Custom labels
    rotation=45,  # Rotating labels
    fontsize=12   # Font size
)
plt.yticks(
    ticks=[10, 20, 30, 40],  # Locations of the ticks
    labels=['Low', 'Medium', 'High', 'Very High'],  # Custom labels
    fontsize=12   # Font size
)

# Adding labels and title
plt.xlabel("Custom X-axis")
plt.ylabel("Custom Y-axis")
plt.title("Example of xticks and yticks")

# Show plot
plt.grid(True)
plt.show()
```
