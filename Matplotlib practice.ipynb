Line Plot

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
x = [1,2,3,4]
y = [10,20,30,40]
plt.plot(x,y)
plt.show()
```

```python
# Line Plot
bastman = pd.read_csv('sharma-kohli.csv')
bastman

# adding a legend
plt.plot(bastman['index'],bastman['V Kohli'],label = 'Virat Kohli')
plt.plot(bastman['index'],bastman['RG Sharma'],label = 'RG Sharma Kohli')

# set location of legend
plt.legend(loc='upper right')

# legend title
plt.legend(title='Bastman Score')

# legend font size and style
plt.legend(fontsize=20)
plt.show()
bastman
```

```python
# label and title
plt.plot(bastman['index'],bastman['V Kohli'])
plt.legend(labels=['Virat Kohli Score'])
plt.title("R Sharma and V Kohli Score")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.show()
```

```python
# color, line(width and style) and marker size
plt.plot(bastman['index'],bastman['V Kohli'],color='g',linestyle='dashdot',linewidth=4,marker='s',ms=10)
plt.legend(labels=['Virat Kohli Score'])
plt.title("R Sharma and V Kohli Score")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.grid()
plt.show()
```

```python
# marker shortcut
plt.plot(bastman['index'],bastman['V Kohli'],'o:g',mec='pink',mfc='b',ms=10)
plt.show()
```

```python
# position of title
plt.plot(bastman['index'],bastman['V Kohli'])
plt.title('Joy', loc='right',color='r',fontsize=20)

# add grid lines

# diisplay only grid lines for x-axis
plt.grid(axis='x')
# diisplay only grid lines for x-axis
# plt.grid(axis='y')

# set line property for the grid
plt.grid(color='r',linestyle='-.',linewidth=2,axis='y')
plt.show()
```

```python
x = [1,2,3,4]
y = [10,20,30,40]
plt.scatter(x,y)
```

```python
df = pd.read_csv('batter.csv')
df = df.head(20)
plt.scatter(df['avg'],df['strike_rate'],color='r')
plt.legend(labels=['Avg','Strike_rate'])
plt.title("Avg and Strike Rate Analysis")
plt.xlabel('Average')
plt.ylabel('Strike Rate')
```

```python
# color and size
plt.scatter(df['avg'],df['strike_rate'],color='r',alpha=0.5)

# color each dot
# colors = ['red','green','black','yellow']
# plt.scatter(df['avg'],df['strike_rate'],c=colors)
plt.show()
```

Bar Plot

```python
childen = [10,20,30,40]
colors = ['green','red','blue','pink']
plt.bar(colors,childen)
plt.show()
```

```python
df = pd.read_csv('batsman_season_record.csv')
df
plt.bar(df['batsman'],df['2015'],width=0.3)
plt.show()
```

```python
# create horizontal bar
colors = ['red','green','blue','pink','black']
plt.barh(df['batsman'],df['2016'],color=colors,height=[0.1,1,2,1,2])
plt.show()
```

```python
plt.bar(df['batsman'],df['2015'])
plt.bar(df['batsman'],df['2016'],bottom=df['2015'])
plt.bar(df['batsman'],df['2017'],bottom=df['2016'])
```

```python
x = ['A','B','C','D']
y = [1,2,3,4]
colors = ['green','blue','red','pink']
heights = [50,20,40,100]
plt.bar(x,y,color = colors,width=0.5)
```

Pie Chart

```python
data = [10,2,20,15,5]
plt.pie(data)
plt.show()
```

```python
df = pd.read_csv('gayle-175.csv')
colors = ['r','g','b','c','y']
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%.1f%%',colors=colors,explode=[0.1,0,0,0,0,0])
plt.show()
df
```

```python
y = [35,25,25,15]
plt.pie(y,labels=['Apple','Banana','Guava','cherry'],autopct='%1.f%%',explode=[0.1,0,0,0],shadow=True)
plt.legend(title="Four Fruits")
plt.show()
```
