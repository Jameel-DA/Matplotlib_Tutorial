```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn
```

Color Scatter Plot

```python
iris = pd.read_csv('iris.csv')
iris.head()
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'])
```

```python
# cmap and alpha
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],cmap='jet',alpha=0.5,c=['r'])
plt.xlabel('Sepal Lenght')
plt.ylabel('Petal Lenght')
plt.colorbar()
plt.show()
```

Plot Size

```python
# we can change the size of plot using figure method
plt.figure(figsize=(15,7)) # 15 is the width of the plot and 7 is height
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],cmap='jet',alpha=0.5,c=['r'])
plt.xlabel('Sepal Lenght')
plt.ylabel('Petal Lenght')
plt.colorbar()
plt.show()
```

Sub plot 

```python
df = df.head()
fig,ax = plt.subplots(nrows=2,ncols=1,figsize=(10,10))
ax[0].scatter(df['avg'],df['strike_rate'])
ax[1].scatter(df['avg'],df['runs'])

ax[0].set_title("avg vs Strike Rate")
ax[0].set_ylabel("strike rate")
ax[0].set_xlabel("Avg")

ax[1].set_title("Avg vs Runs")
ax[1].set_xlabel("avg")
ax[1].set_ylabel("runs")
```

```python
fig,ax = plt.subplots(nrows=2,ncols=2)

ax[0,0]
ax[0,1]
ax[1,0]
ax[1,1]
```

3D Scatter Plots


```python
df = pd.read_csv('batter.csv')
df
```

```python
fig = plt.figure()
ax = plt.subplot(projection='3d')
ax.scatter3D(df['runs'],df['avg'],df['strike_rate'])
ax.set_title("IPL Batsman Analysis")
ax.set_xlabel("Runs")
ax.set_ylabel("Avg")
ax.set_zlabel("Strike Rate")
```

3D Line Plot

```python
x = [0,1,2]
y = [0,5,2]
z = [0,10,9]
fig = plt.figure()
ax = plt.subplot(projection='3d')
ax.plot3D(x,y,z)
```
